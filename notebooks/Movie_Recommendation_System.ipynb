### add the crew column back

```python
movies = pd.read_csv("../data/tmdb_5000_movies.csv")
credits = pd.read_csv("../data/tmdb_5000_credits.csv")

movies = movies.merge(
    credits,
    left_on="id",
    right_on="movie_id"
)

movies = movies.rename(columns={"title_x": "title"})

movies = movies[
    ["id", "title", "overview", "genres", "keywords", "cast", "crew"]
]

movies = movies.dropna(subset=["overview"])
```

### convert crew strings into python lists

```python
movies["crew"] = movies["crew"].apply(ast.literal_eval)
```

### Create a director extraction function

```python
def extract_director(obj):
    directors = []

    for item in obj:
        if item["job"] == "Director":
            directors.append(item["name"])

    return directors
```

### Apply the function

```python
movies["crew"] = movies["crew"].apply(extract_director)
```

## combine all features into one column

### 1.create the tags column

```python
movies["tags"] = (
    movies["overview"] + " " +
    movies["genres"].apply(lambda x: " ".join(x)) + " " +
    movies["keywords"].apply(lambda x: " ".join(x)) + " " +
    movies["cast"].apply(lambda x: " ".join(x)) + " " +
    movies["crew"].apply(lambda x: " ".join(x))
)
```

```python
display(movies[["title", "tags"]].head())
```

### 2.Remove unnecessary columns

```python
movies = movies[["id", "title", "tags"]]
```

## Text Preprocessing and Stemming

### import portstemmer

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
# test stemming
print(ps.stem("loving"))
print(ps.stem("dancing"))
print(ps.stem("playing"))
print(ps.stem("running"))
```

### create a stemming function

```python
def stem_text(text):
    words = []

    for word in text.split():
        words.append(ps.stem(word))

    return " ".join(words)
```

```python
#test the function
print(stem_text("loved loving loves playing played"))
```

### apply stemming

```python
movies["tags"] = movies["tags"].apply(stem_text)
```

```python
display(movies[["title", "tags"]].head())
```

```python
# verify the dataset
print("Dataset shape:", movies.shape)
print("\nFirst movie tags:")
print(movies["tags"].iloc[0])
```

```python
#Import CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
#Create the CountVectorizer object
cv = CountVectorizer(
    max_features=5000,
    stop_words="english"
)
```

```python
#Convert movie tags into numerical vectors
vectors = cv.fit_transform(movies["tags"]).toarray()
```

```python
#check the output
print("Vector shape:", vectors.shape)

print("\nFirst movie vector:")
print(vectors[0])
```

## Calculate Cosine Similarity

```python
#Import the library
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#Calculate similarity
similarity = cosine_similarity(vectors)
```

```python
#Check the result
print("Similarity matrix shape:", similarity.shape)

print("\nSimilarity scores for the first movie:")
print(similarity[0][:10])
```

## Build the movie Recommendation function

### create the recommendation

```python
def recommend(movie):
    movie_index = movies[movies["title"] == movie].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse = True,
        key = lambda x: x[1]
    )[1:6]

    for i in movies_list:
        print(movies.iloc[i[0]].title) 
```

```python
recommend("Iron Man")
```

## Improve the recommendation function

```python
def recommend(movie):
    if movie not in movies["title"].values:
        print("Movie not founf! please check the movie title.")
        return

    movie_index = movies[movies["title"] == movie].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse = True,
        key=lambda x: x[1]
    )[1:6]

    print(f"Movies similar to {movie}:\n")

    for i in movies_list:
        print(movies.iloc[i[0]].title)   
```

```python
recommend("Avatar")
recommend("Iron Man")
recommend("Unknown Movie")
```

## test  the recommendation system


```python
test_movies = [
    "Avatar",
    "Iron Man",
    "Batman Begins",
    "Titanic",
    "The Avengers"
]

for movie in test_movies:
    print("\n---------------------------")
    recommend(movie)
```

```python
import pickle
import numpy as np
import os

similarity = pickle.load(
    open("../models/similarity.pkl", "rb")
)

print("Similarity matrix loaded successfully!")
print("Shape:", similarity.shape)
```

```python
# Store only the top 20 similar movies for each movie

top_similar = {}

for idx in range(similarity.shape[0]):

    scores = similarity[idx]

    similar_indices = np.argsort(scores)[::-1][1:21]

    top_similar[idx] = [
        (int(i), float(scores[i]))
        for i in similar_indices
    ]

print("Top similar movies calculated successfully!")
```
